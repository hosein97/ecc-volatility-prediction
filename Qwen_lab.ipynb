```python

import numpy as np
import pandas as pd
from collections import Counter, defaultdict

datasets = {
    "MAEC-15 Train": kefvp_train_maec15_chunks_qas,
    "MAEC-15 Dev": kefvp_dev_maec15_chunks_qas,
    "MAEC-15 Test": kefvp_test_maec15_chunks_qas,
    "MAEC-16 Train": kefvp_train_maec16_chunks_qas,
    "MAEC-16 Dev": kefvp_dev_maec16_chunks_qas,
    "MAEC-16 Test": kefvp_test_maec16_chunks_qas,
    "EC Test": kefvp_test_ec_chunks_qas,
}

def word_count(text):
    return len(str(text).split()) if text else 0

def analyze_dataset(data, dataset_name):

    # Group chunks by ECC
    eccs = defaultdict(list)

    for chunk_id, chunk_data in data.items():
        ecc_id = str(chunk_id).split("-")[0]
        qas = chunk_data.get("qas", []) or []
        eccs[ecc_id].extend(qas)

    # Dataset-level statistics
    all_qas = []
    chunks_with_qas = 0
    total_chunks = len(data)

    chunk_qa_counts = []
    ecc_qa_counts = []

    question_words = []
    answer_words = []
    combined_words = []
    topics = set()

    for chunk_id, chunk_data in data.items():
        qas = chunk_data.get("qas", []) or []

        chunk_qa_counts.append(len(qas))

        if qas:
            chunks_with_qas += 1

        for qa in qas:
            all_qas.append(qa)

            qw = word_count(qa.get("Q", ""))
            aw = word_count(qa.get("A", ""))

            question_words.append(qw)
            answer_words.append(aw)
            combined_words.append(qw + aw)

            if qa.get("topic"):
                topics.add(qa["topic"])

    for ecc_id, qas in eccs.items():
        ecc_qa_counts.append(len(qas))

    def stats(values):
        if not values:
            return {
                "Mean": 0, "Median": 0,
                "Min": 0, "Max": 0, "Std": 0
            }
        return {
            "Mean": round(np.mean(values), 2),
            "Median": round(np.median(values), 2),
            "Min": int(np.min(values)),
            "Max": int(np.max(values)),
            "Std": round(np.std(values), 2)
        }

    result = {
        "Dataset": dataset_name,
        "ECCs": len(eccs),
        "Chunks": total_chunks,
        "Chunks with QAs": chunks_with_qas,
        "Chunks without QAs": total_chunks - chunks_with_qas,
        "Empty chunk %": round(
            100 * (total_chunks - chunks_with_qas) / total_chunks, 2
        ) if total_chunks else 0,
        "Total QAs": len(all_qas),
        "Unique topics": len(topics),
    }

    # ECC-level statistics
    for key, value in stats(ecc_qa_counts).items():
        result[f"QAs/ECC {key}"] = value

    # Chunk-level statistics
    for key, value in stats(chunk_qa_counts).items():
        result[f"QAs/Chunk {key}"] = value

    # Word-level statistics
    for name, values in [
        ("Question words", question_words),
        ("Answer words", answer_words),
        ("Combined QA words", combined_words),
    ]:
        for key, value in stats(values).items():
            result[f"{name} {key}"] = value

    return result


# Generate summary
summary = pd.DataFrame([
    analyze_dataset(data, name)
    for name, data in datasets.items()
])

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 250)

display(summary.T)
```

```python
from collections import Counter
import pandas as pd

def analyze_topics(data, dataset_name, top_n=30):
    topic_counts = Counter()
    total_qas = 0
    missing_topic = 0

    for chunk in data.values():
        for qa in chunk.get("qas", []):
            total_qas += 1
            topic = qa.get("topic")

            if topic is None or str(topic).strip() == "":
                missing_topic += 1
            else:
                topic_counts[str(topic).strip()] += 1

    rows = [
        {
            "Topic": topic,
            "QA Count": count,
            "Percentage (%)": 100 * count / total_qas
        }
        for topic, count in topic_counts.most_common(top_n)
    ]

    print(f"\n{'=' * 75}")
    print(dataset_name)
    print(f"Total QAs: {total_qas:,}")
    print(f"Number of topics: {len(topic_counts):,}")
    print(
        f"QAs without topic: {missing_topic:,} "
        f"({100 * missing_topic / total_qas:.2f}%)"
        if total_qas else "QAs without topic: 0"
    )

    df = pd.DataFrame(rows)
    display(df.style.format({"Percentage (%)": "{:.2f}%"}))

    return df


# Combine MAEC-15 and MAEC-16 training sets.
# Prefix chunk IDs to prevent overwriting if IDs overlap.
maec_train_combined = {
    **{
        f"MAEC15_{k}": v
        for k, v in datasets["MAEC-15 Train"].items()
    },
    **{
        f"MAEC16_{k}": v
        for k, v in datasets["MAEC-16 Train"].items()
    }
}


# Analyze topics
topic_results = {}

# Combined MAEC training sets
topic_results["MAEC Combined Train"] = analyze_topics(
    maec_train_combined,
    "MAEC-15 + MAEC-16 Combined Training Sets"
)

# Remaining splits analyzed separately
for name in [
    "MAEC-15 Dev",
    "MAEC-15 Test",
    "MAEC-16 Dev",
    "MAEC-16 Test",
    "EC Test"
]:
    topic_results[name] = analyze_topics(
        datasets[name],
        name
    )
```

```python
import numpy as np
import pandas as pd
from collections import defaultdict

def calculate_line_stats(split_dict):
    ecc_data = defaultdict(lambda: {
        "transcript_lines": 0,
        "qa_line_counts": [],
        "total_qas": 0
    })

    for chunk_id, chunk in split_dict.items():
        ecc_id = str(chunk_id).split("-")[0]
        info = ecc_data[ecc_id]

        # Total transcript lines across all chunks of this ECC
        info["transcript_lines"] += len(chunk.get("text_lines", []))

        # Number of source lines for each QA
        for qa in chunk.get("qas", []):
            source_lines = qa.get("source_lines")

            if not source_lines or len(source_lines) < 2:
                continue

            start, end = map(int, source_lines[:2])

            if end >= start:
                info["qa_line_counts"].append(end - start + 1)
                info["total_qas"] += 1

    transcript_lines = [
        x["transcript_lines"] for x in ecc_data.values()
    ]

    qas_per_ecc = [
        x["total_qas"] for x in ecc_data.values()
    ]

    qa_lines_per_ecc = [
        sum(x["qa_line_counts"]) for x in ecc_data.values()
    ]

    all_qa_line_counts = [
        count
        for x in ecc_data.values()
        for count in x["qa_line_counts"]
    ]

    def stats(values):
        return {
            "Mean": np.mean(values) if values else np.nan,
            "Median": np.median(values) if values else np.nan,
            "Std": np.std(values) if values else np.nan,
            "Min": np.min(values) if values else np.nan,
            "Max": np.max(values) if values else np.nan
        }

    result = {}

    for label, values in [
        ("Total transcript lines / ECC", transcript_lines),
        ("Total QAs / ECC", qas_per_ecc),
        ("Total QA source lines / ECC", qa_lines_per_ecc),
        ("Source lines / QA", all_qa_line_counts)
    ]:
        for stat, value in stats(values).items():
            result[f"{label} - {stat}"] = value

    result["Number of ECCs"] = len(ecc_data)
    result["Total QAs"] = sum(qas_per_ecc)

    return result


# Combine all splits within each dataset
maec15_all = {
    f"MAEC15_{k}": v
    for split in ["MAEC-15 Train", "MAEC-15 Dev", "MAEC-15 Test"]
    for k, v in datasets[split].items()
}

maec16_all = {
    f"MAEC16_{k}": v
    for split in ["MAEC-16 Train", "MAEC-16 Dev", "MAEC-16 Test"]
    for k, v in datasets[split].items()
}

ec_test_all = {
    f"EC_{k}": v
    for k, v in datasets["EC Test"].items()
}


# One summary table: datasets as columns, statistics as rows
line_summary = pd.DataFrame({
    "MAEC-15": calculate_line_stats(maec15_all),
    "MAEC-16": calculate_line_stats(maec16_all),
    "EC Test": calculate_line_stats(ec_test_all)
})

print("Transcript and QA Line Statistics")
display(line_summary.round(2))
```

```python
!pip install -q mutagen tqdm

import os
import time
import numpy as np
import pandas as pd

from collections import defaultdict
from mutagen.mp3 import MP3
from tqdm.auto import tqdm


# Cache audio durations to avoid reading the same file repeatedly
audio_duration_cache = {}


def get_audio_duration(path):
    if not os.path.isfile(path):
        return None

    if path not in audio_duration_cache:
        try:
            audio_duration_cache[path] = MP3(path).info.length
        except Exception:
            audio_duration_cache[path] = None

    return audio_duration_cache[path]


def calculate_audio_stats(split_dict, dataset_name):

    start_time = time.time()

    ecc_data = defaultdict(lambda: {
        "audio_files": set(),
        "qa_durations": [],
        "total_qas": 0,
        "qas_with_audio": 0
    })

    # --------------------------------------------------
    # Step 1: Collect audio files and QA source mappings
    # --------------------------------------------------

    print(f"\n[{dataset_name}] Step 1/2: Processing ECCs and QAs")

    for chunk_id, chunk in tqdm(
        split_dict.items(),
        desc=f"{dataset_name}: ECC chunks",
        unit="chunk"
    ):
        ecc_id = str(chunk_id).split("-")[0]
        info = ecc_data[ecc_id]

        audio_map = chunk.get("audio_map", {})
        audio_path = chunk.get("audio_path", "")

        # All audio files belonging to the ECC
        for filename in audio_map.values():
            info["audio_files"].add(
                os.path.join(audio_path, filename)
            )

        # Audio duration corresponding to each QA
        for qa in chunk.get("qas", []):
            info["total_qas"] += 1

            source_lines = qa.get("source_lines")

            if not source_lines or len(source_lines) < 2:
                continue

            start, end = map(int, source_lines[:2])

            if end < start:
                continue

            # Find unique audio files covering the QA's source lines
            qa_files = {
                os.path.join(audio_path, audio_map[str(line)])
                for line in range(start, end + 1)
                if str(line) in audio_map
            }

            durations = [
                get_audio_duration(path)
                for path in qa_files
            ]

            durations = [d for d in durations if d is not None]

            if durations:
                info["qa_durations"].append(sum(durations))
                info["qas_with_audio"] += 1

    # --------------------------------------------------
    # Step 2: Calculate ECC-level audio durations
    # --------------------------------------------------

    print(f"\n[{dataset_name}] Step 2/2: Reading audio durations")

    # Collect unique audio files across all ECCs
    all_audio_files = set()

    for info in ecc_data.values():
        all_audio_files.update(info["audio_files"])

    print(f"Unique audio files to process: {len(all_audio_files):,}")

    # Read durations with progress bar
    for path in tqdm(
        all_audio_files,
        desc=f"{dataset_name}: Audio files",
        unit="file"
    ):
        get_audio_duration(path)

    # --------------------------------------------------
    # Aggregate values across ECCs
    # --------------------------------------------------

    ecc_audio_durations = []
    qa_audio_per_ecc = []
    all_qa_durations = []

    total_audio_files = 0
    total_qas = 0
    total_qas_with_audio = 0

    for info in ecc_data.values():

        file_durations = [
            audio_duration_cache[path]
            for path in info["audio_files"]
            if audio_duration_cache.get(path) is not None
        ]

        ecc_audio_durations.append(sum(file_durations))

        qa_audio_per_ecc.append(sum(info["qa_durations"]))
        all_qa_durations.extend(info["qa_durations"])

        total_audio_files += len(info["audio_files"])
        total_qas += info["total_qas"]
        total_qas_with_audio += info["qas_with_audio"]

    # --------------------------------------------------
    # Summary statistics
    # --------------------------------------------------

    def add_stats(result, label, values):

        values = np.asarray(values, dtype=float)

        stats = [
            ("Mean", np.mean(values) if len(values) else np.nan),
            ("Median", np.median(values) if len(values) else np.nan),
            ("Std", np.std(values) if len(values) else np.nan),
            ("Min", np.min(values) if len(values) else np.nan),
            ("Max", np.max(values) if len(values) else np.nan)
        ]

        for stat, value in stats:
            result[f"{label} - {stat}"] = value

    result = {}

    add_stats(
        result,
        "Total audio duration / ECC (hours)",
        np.array(ecc_audio_durations) / 3600
    )

    add_stats(
        result,
        "Total QA audio duration / ECC (hours)",
        np.array(qa_audio_per_ecc) / 3600
    )

    add_stats(
        result,
        "Audio duration / QA (seconds)",
        all_qa_durations
    )

    result["Number of ECCs"] = len(ecc_data)
    result["Total audio files"] = total_audio_files
    result["Total QAs"] = total_qas
    result["QAs with audio"] = total_qas_with_audio
    result["QAs without audio"] = total_qas - total_qas_with_audio

    elapsed = time.time() - start_time

    print(
        f"\n[{dataset_name}] Completed in "
        f"{elapsed / 60:.1f} minutes."
    )

    return result


# --------------------------------------------------
# Run all three datasets
# --------------------------------------------------

overall_start = time.time()

audio_summary = pd.DataFrame({
    "MAEC-15": calculate_audio_stats(maec15_all, "MAEC-15"),
    "MAEC-16": calculate_audio_stats(maec16_all, "MAEC-16"),
    "EC Test": calculate_audio_stats(ec_test_all, "EC Test")
})

print("\nAudio Duration Statistics")
display(audio_summary.round(2))

print(
    f"\nTotal execution time: "
    f"{(time.time() - overall_start) / 60:.1f} minutes"
)
```

```python
import math
import gc
import pickle
import numpy as np
import torch
import librosa
from tqdm.auto import tqdm

# =========================================================
# CONFIG
# =========================================================
BATCH_SIZE = 4
SAVE_DIR = "./Volatility-Prediction-Data"
os.makedirs(SAVE_DIR, exist_ok=True)

datasets = {
    "kefvp_train_maec15": kefvp_train_maec15_chunks_qas,
    "kefvp_dev_maec15": kefvp_dev_maec15_chunks_qas,
    "kefvp_test_maec15": kefvp_test_maec15_chunks_qas,
    "kefvp_train_maec16": kefvp_train_maec16_chunks_qas,
    "kefvp_dev_maec16": kefvp_dev_maec16_chunks_qas,
    "kefvp_test_maec16": kefvp_test_maec16_chunks_qas,
}

# =========================================================
# HELPER
# =========================================================
def process_batch(batch_items):
    if len(batch_items) == 0:
        return []

    texts = []
    audios = []
    metadata = []

    for item in batch_items:
        chunk_id = item["chunk_id"]
        qa_idx = item["qa_idx"]
        q = item["question"]
        answer = item["answer"]
        topic = item["topic"]
        raw_a = item["raw_a"]
        audio = item["audio"]

        messages = [
            {
                "role": "system",
                "content": (
                    "You are an acoustic and emotional analysis assistant "
                    "specializing in Earnings Conference Calls (ECCs). "
                    "Focus primarily on how the speaker spoke "
                    "(tone, emotion, pace, confidence, hesitation, volatility), "
                    "rather than on the literal words."
                )
            },
            {
                "role": "user",
                "content": [
                    {"type": "audio", "audio": audio},
                    {"type": "text", "text":
                        f"Topic: {topic}\n\n"
                        f"Question: {q}\n\n"
                        f"Answer: {answer}\n\n"
                        f"Transcript:\n{raw_a}\n\n"
                        "Describe the emotional and tonal characteristics of this speech segment."
                    }
                ]
            }
        ]

        text = processor.tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True
        )

        texts.append(text)
        audios.append(audio)

        metadata.append({
            "chunk_id": chunk_id,
            "qa_idx": qa_idx,
            "question": q,
            "topic": topic,
        })

    # =====================================================
    # PROCESS BATCH
    # =====================================================
    inputs = processor(
        text=texts,
        audio=audios,
        sampling_rate=16000,
        return_tensors="pt",
        padding=True
    ).to(model.device)

    with torch.inference_mode():
        outputs = model.thinker(
            **inputs,
            output_hidden_states=True,
            return_dict=True,
        )

    last_hidden = outputs.hidden_states[-1]

    attention_mask = inputs["attention_mask"].unsqueeze(-1)
    masked_hidden = last_hidden * attention_mask

    embedding = masked_hidden.sum(dim=1) / attention_mask.sum(dim=1)

    embedding_np = embedding.detach().float().cpu().numpy()

    results = []
    for i in range(len(metadata)):
        results.append({
            "ecc_chunk_id": metadata[i]["chunk_id"],
            "qa_index": metadata[i]["qa_idx"],
            "question": metadata[i]["question"],
            "topic": metadata[i]["topic"],
            "embedding": embedding_np[i],
        })

    del outputs, last_hidden, embedding, inputs
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return results


# =========================================================
# MAIN LOOP OVER DATASETS
# =========================================================
for dataset_name, dataset in datasets.items():

    print(f"\n========== Processing {dataset_name} ==========")

    save_path = os.path.join(SAVE_DIR, f"{dataset_name}_embeddings.pkl")

    all_embeddings = []
    batch_buffer = []

    chunk_items = list(dataset.items())

    for chunk_id, chunk_data in tqdm(chunk_items, desc=dataset_name):

        qas = chunk_data["qas"]

        for qa_idx, qa in enumerate(qas):
            try:
                q = qa["Q"]
                a = qa["A"]
                topic = qa["topic"]
                source_lines = qa["source_lines"]

                raw_a = ""
                audio_segments = []

                for line in range(source_lines[0], source_lines[1] + 1):
                    raw_a += chunk_data["text_lines"][line - 1] + "\n"

                    audio_file = (
                        chunk_data["audio_path"] + "/" +
                        chunk_data["audio_map"][str(line)]
                    )

                    audio_array_chunk, _ = librosa.load(audio_file, sr=16000)
                    audio_segments.append(audio_array_chunk)

                audio = np.concatenate(audio_segments)

                batch_buffer.append({
                    "chunk_id": chunk_id,
                    "qa_idx": qa_idx,
                    "question": q,
                    "answer": a,
                    "topic": topic,
                    "raw_a": raw_a,
                    "audio": audio,
                })

                if len(batch_buffer) >= BATCH_SIZE:
                    batch_results = process_batch(batch_buffer)
                    all_embeddings.extend(batch_results)
                    batch_buffer = []

            except Exception as e:
                tqdm.write(f"[ERROR] chunk={chunk_id}, qa={qa_idx} -> {e}")
                if torch.cuda.is_available():
                    torch.cuda.empty_cache()
                continue

    # =====================================================
    # FINAL BATCH
    # =====================================================
    if len(batch_buffer) > 0:
        batch_results = process_batch(batch_buffer)
        all_embeddings.extend(batch_results)

    # =====================================================
    # SAVE
    # =====================================================
    with open(save_path, "wb") as f:
        pickle.dump(all_embeddings, f)

    print(f"Saved {len(all_embeddings)} embeddings to {save_path}")
```

```python
import gc
import pickle
import numpy as np
import torch
import librosa
import os
from tqdm.auto import tqdm

# =========================================================
# CONFIG
# =========================================================

BATCH_SIZE = 2

SAVE_DIR = "./Volatility-Prediction-Data"

os.makedirs(SAVE_DIR, exist_ok=True)

datasets = {
    "kefvp_train_maec15": kefvp_train_maec15_chunks_qas,
    "kefvp_dev_maec15": kefvp_dev_maec15_chunks_qas,
    "kefvp_test_maec15": kefvp_test_maec15_chunks_qas,
    "kefvp_train_maec16": kefvp_train_maec16_chunks_qas,
    "kefvp_dev_maec16": kefvp_dev_maec16_chunks_qas,
    "kefvp_test_maec16": kefvp_test_maec16_chunks_qas,
}

# =========================================================
# HELPER
# =========================================================

def process_batch(batch_items):

    if len(batch_items) == 0:
        return []

    texts = []
    audios = []
    metadata = []

    for item in batch_items:

        texts.append(item["text"])
        audios.append(item["audio"])

        metadata.append({
            "ecc_chunk_id": item["chunk_id"]
        })

    # =====================================================
    # PROCESS BATCH
    # =====================================================

    inputs = processor(
        text=texts,
        audio=audios,
        sampling_rate=16000,
        return_tensors="pt",
        padding=True
    ).to(model.device)

    with torch.inference_mode():

        outputs = model.thinker(
            **inputs,
            output_hidden_states=True,
            return_dict=True,
        )

        last_hidden = outputs.hidden_states[-1]

        attention_mask = inputs["attention_mask"].unsqueeze(-1)

        masked_hidden = last_hidden * attention_mask

        embedding = (
            masked_hidden.sum(dim=1)
            / attention_mask.sum(dim=1)
        )

        embedding_np = embedding.detach().float().cpu().numpy()

    results = []

    for i in range(len(metadata)):

        results.append({
            "ecc_chunk_id": metadata[i]["ecc_chunk_id"],
            "embedding": embedding_np[i],
        })

    del outputs
    del last_hidden
    del embedding
    del inputs

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return results


# =========================================================
# MAIN LOOP OVER ALL DATASETS
# =========================================================

for dataset_name, dataset in datasets.items():

    print(f"\n========== Processing {dataset_name} ==========")

    save_path = os.path.join(
        SAVE_DIR,
        f"{dataset_name}_chunk_text_audio_embeddings.pkl"
    )

    all_embeddings = []
    batch_buffer = []

    chunk_items = list(dataset.items())

    for chunk_id, chunk_data in tqdm(
        chunk_items,
        desc=dataset_name
    ):

        try:

            ####################################################
            # BUILD FULL TRANSCRIPT
            ####################################################

            raw_text = "\n".join(chunk_data["text_lines"])

            ####################################################
            # BUILD FULL AUDIO
            ####################################################

            audio_segments = []

            for line in range(1, len(chunk_data["text_lines"]) + 1):

                audio_file = os.path.join(
                    chunk_data["audio_path"],
                    chunk_data["audio_map"][str(line)]
                )

                audio_array, _ = librosa.load(
                    audio_file,
                    sr=16000
                )

                audio_segments.append(audio_array)

            audio = np.concatenate(audio_segments)

            duration_seconds = len(audio) / 16000

            if duration_seconds > 300:
                tqdm.write(
                    f"[WARNING] chunk={chunk_id}: "
                    f"audio duration={duration_seconds:.1f}s "
                    f"({duration_seconds / 60:.1f} minutes)"
                )
            ####################################################
            # ADD TO BATCH — CHAT TEMPLATE
            ####################################################

            messages = [
                {
                    "role": "system",
                    "content": (
                        "You are an acoustic and emotional analysis assistant "
                        "specializing in Earnings Conference Calls (ECCs). "
                        "Focus primarily on how the speaker spoke "
                        "(tone, emotion, pace, confidence, hesitation, volatility), "
                        "rather than on the literal words."
                    )
                },
                {
                    "role": "user",
                    "content": [
                        {"type": "audio", "audio": audio},
                        {
                            "type": "text",
                            "text": (
                                "Analyze the emotional and tonal characteristics "
                                "of this speech segment.\n\n"
                                f"Full transcript:\n{raw_text}"
                            )
                        }
                    ]
                }
            ]

            text = processor.apply_chat_template(
                messages,
                tokenize=False,
                add_generation_prompt=True
            )

            batch_buffer.append({
                "chunk_id": chunk_id,
                "text": text,
                "audio": audio,
            })
            ####################################################
            # PROCESS BATCH
            ####################################################

            if len(batch_buffer) >= BATCH_SIZE:

                batch_results = process_batch(
                    batch_buffer
                )

                all_embeddings.extend(
                    batch_results
                )

                batch_buffer = []

        except Exception as e:

            tqdm.write(
                f"[ERROR] chunk={chunk_id} -> {e}"
            )

            if torch.cuda.is_available():
                torch.cuda.empty_cache()

            continue

    # =========================================================
    # FINAL BATCH
    # =========================================================

    if len(batch_buffer) > 0:

        batch_results = process_batch(
            batch_buffer
        )

        all_embeddings.extend(
            batch_results
        )

    # =========================================================
    # SAVE
    # =========================================================

    with open(save_path, "wb") as f:

        pickle.dump(
            all_embeddings,
            f
        )

    print(
        f"Saved {len(all_embeddings)} embeddings to {save_path}"
    )
```

```python
from pathlib import Path
import pickle

DATA_DIR = Path("/content/Volatility-Prediction-Data/volatility_embeddings")

MODALITY_SUFFIX = {
    "text_audio_prompt1": "embeddings",
    "text_only": "text_embeddings",
    "text_audio_prompt2": "qa_text_audio_prompt2_embeddings",
    "audio_only": "audio_embeddings",
    "chunk_text_audio": "chunk_text_audio_embeddings",
}

MODALITIES = list(MODALITY_SUFFIX)


def load_embeddings(dataset: str, split: str, modality: str):
    suffix = MODALITY_SUFFIX[modality]
    fname = f"kefvp_{split}_{dataset}_{suffix}.pkl"
    path = DATA_DIR / fname

    with open(path, "rb") as f:
        return pickle.load(f)


EMBEDDINGS = {}

for dataset in ("maec15", "maec16"):
    for split in ("train", "dev", "test"):
        for modality in MODALITIES:
            EMBEDDINGS[(dataset, split, modality)] = load_embeddings(
                dataset, split, modality
            )

for modality in MODALITIES:
    EMBEDDINGS[("ec", "test", modality)] = load_embeddings(
        "ec", "test", modality
    )

print(f"Loaded {len(EMBEDDINGS)} embedding files.")
```

```python
for key, items in EMBEDDINGS.items():
    if key[0] not in ("maec15", "maec16"):
        continue

    for item in items:
        idx = int(item["ecc_chunk_id"].split("-")[0])
        row = maec_transcripts_df.iloc[idx]

        item["date"] = str(row["date"])
        item["ticker"] = row["ticker"]

print("Added date and ticker to all MAEC15 and MAEC16 embeddings.")
```

```python
for key, items in EMBEDDINGS.items():
    if key[0] != "ec":
        continue

    for item in items:
        idx = int(item["ecc_chunk_id"].split("-")[0])
        row = ec_transcripts_test_df.iloc[idx]

        item["date"] = str(row["date"])
        item["ticker"] = row["ticker"]

print("Added date and ticker to all EC embeddings.")
```

```python
import pickle

for (dataset, split, modality), items in EMBEDDINGS.items():
    suffix = MODALITY_SUFFIX[modality]
    filename = f"kefvp_{split}_{dataset}_{suffix}.pkl"
    filepath = DATA_DIR / filename

    with open(filepath, "wb") as f:
        pickle.dump(items, f)

    print(f"Saved {filepath}")

print("Done.")
```
