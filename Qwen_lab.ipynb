```python

from transformers import (
    Qwen2_5OmniForConditionalGeneration,
    Qwen2_5OmniProcessor,
    BitsAndBytesConfig
)

import torch
import librosa

model_name = "Qwen/Qwen2.5-Omni-7B"

# ==========================================
# Quantization config
# ==========================================

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,

    # best 4bit quant type
    bnb_4bit_quant_type="nf4",

    # compute in bf16 on GPU
    bnb_4bit_compute_dtype=torch.bfloat16,

    # extra compression
    bnb_4bit_use_double_quant=True,
)

# ==========================================
# Load quantized model
# ==========================================

model = Qwen2_5OmniForConditionalGeneration.from_pretrained(
    model_name,

    quantization_config=bnb_config,

    device_map="auto",

    trust_remote_code=True,

 )

processor = Qwen2_5OmniProcessor.from_pretrained(
    model_name,
    trust_remote_code=True
)

model.eval()
```

```python
import math
import gc
import pickle
import numpy as np
import torch
import librosa
from tqdm.auto import tqdm

# =========================================================
# CONFIG
# =========================================================
BATCH_SIZE = 4
SAVE_DIR = "./Volatility-Prediction-Data"
os.makedirs(SAVE_DIR, exist_ok=True)

datasets = {
    "kefvp_train_maec15": kefvp_train_maec15_chunks_qas,
    "kefvp_dev_maec15": kefvp_dev_maec15_chunks_qas,
    "kefvp_test_maec15": kefvp_test_maec15_chunks_qas,
    "kefvp_train_maec16": kefvp_train_maec16_chunks_qas,
    "kefvp_dev_maec16": kefvp_dev_maec16_chunks_qas,
    "kefvp_test_maec16": kefvp_test_maec16_chunks_qas,
}

# =========================================================
# HELPER
# =========================================================
def process_batch(batch_items):
    if len(batch_items) == 0:
        return []

    texts = []
    audios = []
    metadata = []

    for item in batch_items:
        chunk_id = item["chunk_id"]
        qa_idx = item["qa_idx"]
        q = item["question"]
        answer = item["answer"]
        topic = item["topic"]
        raw_a = item["raw_a"]
        audio = item["audio"]

        messages = [
            {
                "role": "system",
                "content": (
                    "You are an acoustic and emotional analysis assistant "
                    "specializing in Earnings Conference Calls (ECCs). "
                    "Focus primarily on how the speaker spoke "
                    "(tone, emotion, pace, confidence, hesitation, volatility), "
                    "rather than on the literal words."
                )
            },
            {
                "role": "user",
                "content": [
                    {"type": "audio", "audio": audio},
                    {"type": "text", "text":
                        f"Topic: {topic}\n\n"
                        f"Question: {q}\n\n"
                        f"Answer: {answer}\n\n"
                        f"Transcript:\n{raw_a}\n\n"
                        "Describe the emotional and tonal characteristics of this speech segment."
                    }
                ]
            }
        ]

        text = processor.tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True
        )

        texts.append(text)
        audios.append(audio)

        metadata.append({
            "chunk_id": chunk_id,
            "qa_idx": qa_idx,
            "question": q,
            "topic": topic,
        })

    # =====================================================
    # PROCESS BATCH
    # =====================================================
    inputs = processor(
        text=texts,
        audio=audios,
        sampling_rate=16000,
        return_tensors="pt",
        padding=True
    ).to(model.device)

    with torch.inference_mode():
        outputs = model.thinker(
            **inputs,
            output_hidden_states=True,
            return_dict=True,
        )

    last_hidden = outputs.hidden_states[-1]

    attention_mask = inputs["attention_mask"].unsqueeze(-1)
    masked_hidden = last_hidden * attention_mask

    embedding = masked_hidden.sum(dim=1) / attention_mask.sum(dim=1)

    embedding_np = embedding.detach().float().cpu().numpy()

    results = []
    for i in range(len(metadata)):
        results.append({
            "ecc_chunk_id": metadata[i]["chunk_id"],
            "qa_index": metadata[i]["qa_idx"],
            "question": metadata[i]["question"],
            "topic": metadata[i]["topic"],
            "embedding": embedding_np[i],
        })

    del outputs, last_hidden, embedding, inputs
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return results


# =========================================================
# MAIN LOOP OVER DATASETS
# =========================================================
for dataset_name, dataset in datasets.items():

    print(f"\n========== Processing {dataset_name} ==========")

    save_path = os.path.join(SAVE_DIR, f"{dataset_name}_embeddings.pkl")

    all_embeddings = []
    batch_buffer = []

    chunk_items = list(dataset.items())

    for chunk_id, chunk_data in tqdm(chunk_items, desc=dataset_name):

        qas = chunk_data["qas"]

        for qa_idx, qa in enumerate(qas):
            try:
                q = qa["Q"]
                a = qa["A"]
                topic = qa["topic"]
                source_lines = qa["source_lines"]

                raw_a = ""
                audio_segments = []

                for line in range(source_lines[0], source_lines[1] + 1):
                    raw_a += chunk_data["text_lines"][line - 1] + "\n"

                    audio_file = (
                        chunk_data["audio_path"] + "/" +
                        chunk_data["audio_map"][str(line)]
                    )

                    audio_array_chunk, _ = librosa.load(audio_file, sr=16000)
                    audio_segments.append(audio_array_chunk)

                audio = np.concatenate(audio_segments)

                batch_buffer.append({
                    "chunk_id": chunk_id,
                    "qa_idx": qa_idx,
                    "question": q,
                    "answer": a,
                    "topic": topic,
                    "raw_a": raw_a,
                    "audio": audio,
                })

                if len(batch_buffer) >= BATCH_SIZE:
                    batch_results = process_batch(batch_buffer)
                    all_embeddings.extend(batch_results)
                    batch_buffer = []

            except Exception as e:
                tqdm.write(f"[ERROR] chunk={chunk_id}, qa={qa_idx} -> {e}")
                if torch.cuda.is_available():
                    torch.cuda.empty_cache()
                continue

    # =====================================================
    # FINAL BATCH
    # =====================================================
    if len(batch_buffer) > 0:
        batch_results = process_batch(batch_buffer)
        all_embeddings.extend(batch_results)

    # =====================================================
    # SAVE
    # =====================================================
    with open(save_path, "wb") as f:
        pickle.dump(all_embeddings, f)

    print(f"Saved {len(all_embeddings)} embeddings to {save_path}")
```

```python
### MAIN SCRIPT TO GENERATE AND SAVE EMBEDDINGS (BATCHED)

import math
import gc
import pickle
import numpy as np
import torch
import librosa
import os
from tqdm.auto import tqdm

# =========================================================
# CONFIG
# =========================================================

BATCH_SIZE = 4

SAVE_DIR = "./Volatility-Prediction-Data"

os.makedirs(SAVE_DIR, exist_ok=True)

datasets = {
    "kefvp_train_maec15": kefvp_train_maec15_chunks_qas,
    "kefvp_dev_maec15": kefvp_dev_maec15_chunks_qas,
    "kefvp_test_maec15": kefvp_test_maec15_chunks_qas,
    "kefvp_train_maec16": kefvp_train_maec16_chunks_qas,
    "kefvp_dev_maec16": kefvp_dev_maec16_chunks_qas,
    "kefvp_test_maec16": kefvp_test_maec16_chunks_qas,
}

# =========================================================
# HELPER
# =========================================================

def process_batch(batch_items):

    if len(batch_items) == 0:
        return []

    texts = []
    metadata = []

    # =====================================================
    # BUILD INPUTS
    # =====================================================

    for item in batch_items:

        chunk_id = item["chunk_id"]
        qa_idx = item["qa_idx"]
        q = item["question"]
        answer = item["answer"]
        topic = item["topic"]
        raw_a = item["raw_a"]

        messages = [
            {
                "role": "system",
                "content": (
                    "You are an analysis assistant specializing in "
                    "Earnings Conference Calls (ECCs)."
                )
            },
            {
                "role": "user",
                "content": (
                    f"Topic: {topic}\n\n"
                    f"Question: {q}\n\n"
                    f"Answer: {answer}\n\n"
                    f"Transcript:\n{raw_a}\n\n"
                    "Describe the characteristics of this answer."
                )
            }
        ]

        text = processor.tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True
        )

        texts.append(text)

        metadata.append({
            "ecc_chunk_id": chunk_id,
            "qa_index": qa_idx,
            "question": q,
            "topic": topic,
        })

    # =====================================================
    # PROCESS BATCH
    # =====================================================

    inputs = processor(
        text=texts,
        return_tensors="pt",
        padding=True
    ).to(model.device)

    with torch.inference_mode():

        outputs = model.thinker(
            **inputs,
            output_hidden_states=True,
            return_dict=True,
        )

        last_hidden = outputs.hidden_states[-1]

        attention_mask = inputs["attention_mask"]
        mask = attention_mask.unsqueeze(-1)

        masked_hidden = last_hidden * mask

        embedding = masked_hidden.sum(dim=1) / mask.sum(dim=1)

        embedding_np = embedding.detach().float().cpu().numpy()

    # =====================================================
    # UNPACK RESULTS
    # =====================================================

    results = []

    for i in range(len(metadata)):

        results.append({
            "ecc_chunk_id": metadata[i]["ecc_chunk_id"],
            "qa_index": metadata[i]["qa_index"],
            "question": metadata[i]["question"],
            "topic": metadata[i]["topic"],
            "embedding": embedding_np[i],
        })

    # =====================================================
    # CLEANUP
    # =====================================================

    del outputs, last_hidden, embedding, inputs
    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return results


# =========================================================
# MAIN LOOP OVER ALL DATASETS
# =========================================================

for dataset_name, dataset in datasets.items():

    print(f"\n========== Processing {dataset_name} ==========")

    save_path = os.path.join(SAVE_DIR, f"{dataset_name}_text_embeddings.pkl")

    all_embeddings = []
    batch_buffer = []

    chunk_items = list(dataset.items())

    for local_chunk_idx, (chunk_id, chunk_data) in enumerate(
        tqdm(chunk_items, desc=dataset_name)
    ):

        qas = chunk_data["qas"]

        for qa_idx, qa in enumerate(qas):

            try:

                q = qa["Q"]
                a = qa["A"]
                topic = qa["topic"]
                source_lines = qa["source_lines"]

                raw_a = ""

                # =====================================================
                # BUILD TEXT ONLY (NO AUDIO HERE)
                # =====================================================

                for line in range(source_lines[0], source_lines[1] + 1):
                    raw_a += chunk_data["text_lines"][line - 1] + "\n"

                # =====================================================
                # ADD TO BATCH BUFFER
                # =====================================================

                batch_buffer.append({
                    "chunk_id": chunk_id,
                    "qa_idx": qa_idx,
                    "question": q,
                    "answer": a,
                    "topic": topic,
                    "raw_a": raw_a,
                })

                # =====================================================
                # PROCESS FULL BATCH
                # =====================================================

                if len(batch_buffer) >= BATCH_SIZE:

                    batch_results = process_batch(batch_buffer)

                    all_embeddings.extend(batch_results)

                    batch_buffer = []

            except Exception as e:

                tqdm.write(
                    f"\n[ERROR] chunk={chunk_id} qa={qa_idx} -> {e}"
                )

                if torch.cuda.is_available():
                    torch.cuda.empty_cache()

                continue

    # =====================================================
    # FINAL BATCH
    # =====================================================

    if len(batch_buffer) > 0:

        batch_results = process_batch(batch_buffer)

        all_embeddings.extend(batch_results)

    # =====================================================
    # SAVE
    # =====================================================

    with open(save_path, "wb") as f:
        pickle.dump(all_embeddings, f)

    print(f"Saved {len(all_embeddings)} embeddings to {save_path}")
```

```python
import math
import gc
import pickle
import numpy as np
import torch
import librosa
from tqdm.auto import tqdm

# =========================================================
# CONFIG
# =========================================================
BATCH_SIZE = 4
SAVE_DIR = "./Volatility-Prediction-Data"

os.makedirs(SAVE_DIR, exist_ok=True)

datasets = {
    "kefvp_test_ec": kefvp_test_ec_chunks_qas,
}

# =========================================================
# HELPER
# =========================================================
def process_batch(batch_items):
    if len(batch_items) == 0:
        return []

    texts = []
    audios = []
    metadata = []

    for item in batch_items:
        chunk_id = item["chunk_id"]
        qa_idx = item["qa_idx"]
        q = item["question"]
        topic = item["topic"]
        audio = item["audio"]

        # -------------------------------------------------
        # Audio ONLY
        # -------------------------------------------------
        messages = [
            {
                "role": "system",
                "content": (
                    "You are an acoustic analysis assistant. "
                    "Focus only on the acoustic properties of the speech."
                ),
            },
            {
                "role": "user",
                "content": [
                    {
                        "type": "audio",
                        "audio": audio,
                    }
                ],
            },
        ]

        text = processor.tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True,
        )

        texts.append(text)
        audios.append(audio)

        metadata.append({
            "chunk_id": chunk_id,
            "qa_idx": qa_idx,
            "question": q,
            "topic": topic,
        })

    # -------------------------------------------------
    # Processor
    # -------------------------------------------------
    inputs = processor(
        text=texts,
        audio=audios,
        sampling_rate=16000,
        return_tensors="pt",
        padding=True,
    ).to(model.device)

    with torch.inference_mode():
        outputs = model.thinker(
            **inputs,
            output_hidden_states=True,
            return_dict=True,
        )

    last_hidden = outputs.hidden_states[-1]

    attention_mask = inputs["attention_mask"].unsqueeze(-1)
    masked_hidden = last_hidden * attention_mask

    embedding = masked_hidden.sum(dim=1) / attention_mask.sum(dim=1)

    embedding_np = embedding.detach().float().cpu().numpy()

    results = []
    for i in range(len(metadata)):
        results.append({
            "ecc_chunk_id": metadata[i]["chunk_id"],
            "qa_index": metadata[i]["qa_idx"],
            "question": metadata[i]["question"],
            "topic": metadata[i]["topic"],
            "embedding": embedding_np[i],
        })

    del outputs, last_hidden, embedding, inputs
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return results


# =========================================================
# MAIN LOOP OVER DATASETS
# =========================================================
for dataset_name, dataset in datasets.items():

    print(f"\n========== Processing {dataset_name} ==========")

    save_path = os.path.join(SAVE_DIR, f"{dataset_name}_audio_embeddings.pkl")

    all_embeddings = []
    batch_buffer = []

    chunk_items = list(dataset.items())

    for chunk_id, chunk_data in tqdm(chunk_items, desc=dataset_name):

        qas = chunk_data["qas"]

        for qa_idx, qa in enumerate(qas):
            try:
                q = qa["Q"]
                topic = qa["topic"]
                source_lines = qa["source_lines"]

                raw_a = ""
                audio_segments = []

                for line in range(source_lines[0], source_lines[1] + 1):
                    raw_a += chunk_data["text_lines"][line - 1] + "\n"

                    audio_file = (
                        chunk_data["audio_path"] + "/" +
                        chunk_data["audio_map"][str(line)]
                    )

                    audio_array_chunk, _ = librosa.load(audio_file, sr=16000)
                    audio_segments.append(audio_array_chunk)

                audio = np.concatenate(audio_segments)

                batch_buffer.append({
                    "chunk_id": chunk_id,
                    "qa_idx": qa_idx,
                    "question": q,
                    "topic": topic,
                    "raw_a": raw_a,
                    "audio": audio,
                })

                if len(batch_buffer) >= BATCH_SIZE:
                    batch_results = process_batch(batch_buffer)
                    all_embeddings.extend(batch_results)
                    batch_buffer = []

            except Exception as e:
                tqdm.write(f"[ERROR] chunk={chunk_id}, qa={qa_idx} -> {e}")
                if torch.cuda.is_available():
                    torch.cuda.empty_cache()
                continue

    # =====================================================
    # FINAL BATCH
    # =====================================================
    if len(batch_buffer) > 0:
        batch_results = process_batch(batch_buffer)
        all_embeddings.extend(batch_results)

    # =====================================================
    # SAVE
    # =====================================================
    with open(save_path, "wb") as f:
        pickle.dump(all_embeddings, f)

    print(f"Saved {len(all_embeddings)} embeddings to {save_path}")
```

```python
from pathlib import Path
import pickle

DATA_DIR = Path("/content/Volatility-Prediction-Data/volatility_embeddings")

MODALITY_SUFFIX = {
    "text_audio_prompt1": "embeddings",
    "text_only": "text_embeddings",
    "text_audio_prompt2": "qa_text_audio_prompt2_embeddings",
    "audio_only": "audio_embeddings",
    "chunk_text_audio": "chunk_text_audio_embeddings",
}

MODALITIES = list(MODALITY_SUFFIX)


def load_embeddings(dataset: str, split: str, modality: str):
    suffix = MODALITY_SUFFIX[modality]
    fname = f"kefvp_{split}_{dataset}_{suffix}.pkl"
    path = DATA_DIR / fname

    with open(path, "rb") as f:
        return pickle.load(f)


EMBEDDINGS = {}

for dataset in ("maec15", "maec16"):
    for split in ("train", "dev", "test"):
        for modality in MODALITIES:
            EMBEDDINGS[(dataset, split, modality)] = load_embeddings(
                dataset, split, modality
            )

for modality in MODALITIES:
    EMBEDDINGS[("ec", "test", modality)] = load_embeddings(
        "ec", "test", modality
    )

print(f"Loaded {len(EMBEDDINGS)} embedding files.")
```

```python
for key, items in EMBEDDINGS.items():
    if key[0] not in ("maec15", "maec16"):
        continue

    for item in items:
        idx = int(item["ecc_chunk_id"].split("-")[0])
        row = maec_transcripts_df.iloc[idx]

        item["date"] = str(row["date"])
        item["ticker"] = row["ticker"]

print("Added date and ticker to all MAEC15 and MAEC16 embeddings.")
```

```python
import pickle

for (dataset, split, modality), items in EMBEDDINGS.items():
    suffix = MODALITY_SUFFIX[modality]
    filename = f"kefvp_{split}_{dataset}_{suffix}.pkl"
    filepath = DATA_DIR / filename

    with open(filepath, "wb") as f:
        pickle.dump(items, f)

    print(f"Saved {filepath}")

print("Done.")
```
